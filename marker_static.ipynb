```python
import os
from collections import Counter

# 파일 탐색기에 표시된 상위 경로
# (만약 경로를 찾을 수 없다고 나오면 절대경로인 "/home/jupyter/image_team/exp3_inbox" 등으로 수정해 주세요)
base_dir = "/home/jupyter/data/image_team/exp3_inbox" 

suffix_counts = Counter()

if os.path.exists(base_dir):
    for root_folder in os.listdir(base_dir):
        root_path = os.path.join(base_dir, root_folder)
        
        # 'S'로 시작하는 폴더인지 확인
        if os.path.isdir(root_path) and root_folder.startswith('S'):
            try:
                # 하위 폴더 탐색
                for sub_folder in os.listdir(root_path):
                    sub_path = os.path.join(root_path, sub_folder)
                    
                    # 하위 항목이 폴더이고 '-'가 포함되어 있는지 확인 (예: S1234567-CD68)
                    if os.path.isdir(sub_path) and '-' in sub_folder:
                        # '-' 기준으로 맨 뒤 문자열 추출
                        suffix = sub_folder.split('-')[-1]
                        suffix_counts[suffix] += 1
            except PermissionError:
                pass
                
    # 통계 출력 (개수가 많은 순서대로)
    print("📊 하위 폴더명 (마커) 통계")
    print("-" * 30)
    if suffix_counts:
        for suffix, count in suffix_counts.most_common():
            print(f"{suffix}: {count}개")
    else:
        print("조건에 맞는 하위 폴더를 찾지 못했습니다.")
else:
    print(f"경로를 찾을 수 없습니다. base_dir을 확인해 주세요: {base_dir}")
```
